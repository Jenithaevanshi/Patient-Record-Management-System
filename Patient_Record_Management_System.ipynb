```python
patients = []
appointments = []
bills = []
```

```python
# Add patients
def add_patient():
    patient_id = input("Enter the patient_ID:")
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient id already exists")
            return
    name = input("Enter the patient name:")
    age = input("Enter the Age:")
    gender = input("Enter the gender:")
    phone = input("Enter the phone number:")
    disease = input("enter the disease:")
    doctor = input("enter the doctor name:")
    patient = {
        "id" : patient_id,
        "name": name,
        "age" : age,
        "gender":gender,
        "phone":phone,
        "disease": disease,
        "doctor": doctor
        }
    patients.append(patient)
    print("\n Patients added successfully.")
```

```python
add_patient()
```

```python
#view patient
def view_patients():
    if len(patients)==0:
        print("No patient records found.")
        return
    for patient in patients:
        print("-------------------")
        print("patient_ID:",patient["id"])
        print("Name:",patient["name"])
        print("Age:",patient["age"])
        print("Gender:",patient["gender"])
        print("Phone:",patient["phone"])
        print("Disease:",patient["disease"])
        print("Doctor:",patient["doctor"])
```

```python
view_patients()
```

```python
# search patient
def search_patient():
    patient_id = input("Enter patient ID:")
    found = False
    for patient in patients:
        if patient["id"]==patient_id:
            print("patient_ID:",patient["id"])
            print("Name:",patient["name"])
            print("Age:",patient["age"])
            print("Gender:",patient["gender"])
            print("Phone:",patient["phone"])
            print("Disease:",patient["disease"])
            print("Doctor:",patient["doctor"])
            found = True
            break
    if not found:
        print("Patient not found")
            
    
```

```python
search_patient()
```

```python
# update 
def update_patient():
    patient_id = input("Enter the patient ID:")
    for patient in patients:
        if patient["id"] == patient_id:
            print("\n patient found")
            print("1. update name:")
            print("2. update age:")
            print("3. update gender:")
            print("4. update phone:")
            print("5. update disease:")
            print("6. update doctor:")
            choice = input("Enter your choice:")
            if choice == "1":
                patient["name"]= input("Enter the new name:")
            elif choice =="2":
                patient["age"]= input("Enter the new age:")
            elif choice == "3":
                patient["gender"] = input("Enter the new gender:")
            elif choice == "4":
                patient["phone"] = input("Enter the new phone:")
            elif choice == "5":
                patient["disease"]=input("enter the new disease:")
            elif choice == "6":
                patient["doctor"]= input("enter the new doctor")
            else: 
                print("Invalid choice.")
                return
            print("\n Patient updated successfully.")
            return
    print("Patient record not found.")
```

```python
update_patient()
```

```python
# delete 
def delete_patient():
    patient_id = input("Enter patient Id:")
    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("patient deleted successfully.")
            return
    print("Patient not found")
```

```python
delete_patient()
```

```python
# book appointment
def book_appointment():
    patient_id = input("Enter patient Id:")
    patient_found = False
    for patient in patients:
        if patient["id"] == patient_id:
            patient_found = True 
            patient_name = patient["name"]
            break      

    if not patient_found:
        print("patient not found please add the patient first")
        return
    date = input("Enter the appointment date:")
    time = input("Enter the appointment time:")
    doctor= input("Enter the doctor name:")
    appointment = {
        "patient_id" : patient_id,
        "patient_name": patient_name,
        "date": date,
        "time": time,
        "doctor": doctor
    }
    appointments.append(appointment)
    print("\n Appointments booked successfully.")
```

```python
book_appointment()
```

```python
# view appointment
def view_appointments():
    if len(appointments)==0:
        print("No appointments available.")
        return
    for appointment in appointments:
        print("Patient Id:",appointment["patient_id"])
        print("Patient Name:", appointment["patient_name"])
        print("Date:", appointment["date"])
        print("Time:",appointment["time"])
        print("Doctor:", appointment["doctor"])
```

```python
view_appointments()
```

```python
# generate bills
def generate_bills():
    patient_id = input("Enter the patient ID:")
    patient_found = False
    for patient in patients:
        if patient["id"]==patient_id:
            patient_found = True
            patient_name = patient["name"]
            break
            
        
    if not patient_found:
        print("patient not found")
        return


    consultation = float(input("Enter the consultation amount"))
    medicine = float(input("Enter the medicine amount"))
    test = float(input("Enter the test amount"))
    total = consultation + medicine + test
    bill ={
        "patient_id": patient_id,
        "patient_name": patient_name,
        "consultation": consultation,
        "medicine": medicine,
        "test": test,
        "total": total
    }
    bills.append(bill)
    print("---------")
    print("patient_id:", patient_id)
    print("patient_name:",patient_name)
    print("consultation amount:",consultation)
    print("medicine amount:",medicine)
    print("test amount:",test)
    print("total amount:",total)
```

```python
generate_bills()
```

```python
def view_bills():
    if len(bills) == 0:
        print("No bills available")
        return
    for bill in bills:
        print("patient_id:", bill["patient_id"])
        print("patient_name:",bill["patient_name"])
        print("consultation fee:", bill["consultation"])
        print("medicine fee:",bill["medicine"])
        print("test fee:",bill["test"])
        print("total charges:",bill["total"])
```

```python
while True:
    print("Patient record management system")
    print("1. Add Patients")
    print("2. View patients")
    print("3. Search patients")
    print("4. Update patients")
    print("5. Delete patients")
    print("6. Book appointments")
    print("7. view Appointments")
    print("8. Generate bills")
    print("9. View bills")
    print("10. Exit")
    choice = input("\n Enter your choice")
    if choice == "1":
        add_patient()
    elif choice == "2":
        view_patients()
    elif choice == "3":
        search_patient()
    elif choice == "4":
        update_patient()
    elif choice == "5":
        delete_patient()
    elif choice == "6":
        book_appointment()
    elif choice == "7":
        view_appointments()
    elif choice == "8":
        generate_bills()
    elif choice == "9":
        view_bills()
    elif choice == "10":
        print("Thank you")
        break
    else:
        print("Invalid choice")
```
